**Dictionary**
A dictionary is an unordered, mutable collection of key–value pairs.

student = {
    "name": "Anchal",
    "age": 18,
    "branch": "IT"
}


Think of it like a real dictionary:
word → meaning, key → value

🔹 Key Properties

✔ Stores data as key : value
✔ Keys must be unique
✔ Keys must be immutable (int, str, tuple)
✔ Values can be anything
✔ Mutable (can change values)

```python
d1 = {}
print(type(d1))

d2 = dict({})
print(d2)

d3 = {
    "name" : "Anchal",
    "age": "19",
    "branch": "IT"
}
print(d3)

d4 = dict(name = "Anchal", age = "19", branch = "IT")
print(d4)

d5 = dict([
    ("name", "Anchal"),
    ("age", "19"),
    ("branch", "IT")
])
print(d5)
```

```python
#Accessing Value and manipulating them
d3 = {
    "name" : "Anchal",
    "age": "19",
    "branch": "IT"
}
print(d3)
print(d3["name"])
print(d3["age"])
print(d3["branch"])

d3["name"] = "Shubhii"
d3["age"] = "20"
d3["branch"] = "CSE"
print(d3)

print(d3.get("name"))
print(d3.get("age"))
print(d3.get("branch"))

print(d3.pop("name"))# selects the item to be removed
print(d3)

print(d3.popitem())# delets from the end
print(d3)

car = {
    "brand": "Ford",
    "model": "Mustabng",
    "year": 1964
}
print(car)
print(car.keys())
print(car.values())
print(car.items())

#Looping through Dictionary
for key in car:
  print(key , car[key])

for key, value in car.items():
  print(key, value)



```

```python
squares = {x:x*x for x in range(1,6)}
print(squares)

#Nested Dictionary

student = {
    "name": "Anchal",
    "age": 18,
    "branch": "IT",
    "marks":{
        "maths": 90,
        "english" : 85,
        "science" : 95
    }
}
student.update({"Year":2023-27})
print(student)
print(student['marks']["english"])

#Duplicate keys

d = {
    'a': 20,
    'a': 30,
    'a': 40
}
print(d)

d2 = {
    (1,2) : "point"
}
print(d2)

d1= {
    [1,2] : "hello"
    # unhashable type: 'list'
}
print(d1)
```

```python
#Frequency count
nums = [1,1,3,4, 3, 2 , 4]
freq = {}

for i in nums:
  freq[i] = freq.get(i,0) + 1
print(freq)

#Word Count
sentence = "Python is  easy  python is powerful."
words = sentence.split();

count ={}

for w in words:
  count[w] = count.get(w,0) + 1;

print(count)

```
